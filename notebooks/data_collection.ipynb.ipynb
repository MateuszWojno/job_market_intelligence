# Job Market Intelligence – Data Collection

This notebook runs the data collection pipeline for No Fluff Jobs.

The scraping logic is implemented in the `src/nfj` package.
Raw datasets are stored in `data/raw/` and excluded from version control.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
SRC_PATH = PROJECT_ROOT / "src"

if str(SRC_PATH) not in sys.path:
    sys.path.insert(0, str(SRC_PATH))

from nfj import (
    collect_job_urls,
    scrape_jobs,
)

from nfj.config import (
    URLS_PATH,
    OUTPUT_PATH,
)

print("PROJECT_ROOT:", PROJECT_ROOT)
print("Imports successful.")
```

## 1. Collect job URLs

Collect unique job-offer URLs from No Fluff Jobs categories.

```python
RUN_URL_COLLECTION = False

if RUN_URL_COLLECTION:
    df_urls = collect_job_urls()
else:
    print("URL collection skipped.")
```

## 2. Scrape job offers

Scrape detailed information for job offers that have not been collected yet.

```python
RUN_SCRAPING = False

if RUN_SCRAPING:
    df_jobs = scrape_jobs(
        delay=3.0,
        headless=False,
        max_consecutive_page_errors=5,
    )
else:
    print("Job scraping skipped.")
```

## 3. Validate collected data

Check dataset completeness, uniqueness, and invalid job pages.

```python
df_urls = pd.read_csv(URLS_PATH)
df_jobs = pd.read_csv(OUTPUT_PATH)

bad_mask = (
    df_jobs["title"]
    .fillna("")
    .str.contains(
        "Ta strona nie działa",
        case=False,
        regex=False
    )
)

print("URLs:", len(df_urls))
print("Unique URLs:", df_urls["url"].nunique())

print()
print("Jobs:", len(df_jobs))
print("Unique jobs:", df_jobs["url"].nunique())

print()
print("Invalid jobs:", bad_mask.sum())

print(
    "Missing URLs:",
    len(set(df_urls["url"]) - set(df_jobs["url"]))
)
```
